# 02 — Загрузка canonical panel

Загружает сырые данные для одного рынка и сохраняет их в стандартизованный формат (canonical panel) на диск.

**Что делает:**
- Читает YAML-конфиг рынка через `load_market_config` (по умолчанию `"usa"`)
- Запускает соответствующий загрузчик (`get_loader`) — парсит ZIP-архив, выравнивает индекс, отбирает нужные колонки
- Сохраняет результат в `config.canonical_path` через `save_canonical_panel`
- Выводит форму итогового датафрейма цен как быструю проверку

**Когда использовать:** один раз после получения нового архива, либо при смене параметров конфига загрузчика. Результат используется всеми downstream-ноутбуками.

```python
from sarb.config import load_market_config
from sarb.data import build_canonical_panel, save_canonical_panel
from sarb.data.loaders import get_loader

market_name = "usa"
config = load_market_config(market_name)
panel = get_loader(config).load()
save_canonical_panel(panel, config.canonical_path)
panel.prices.shape
```
