# 07 — Сравнение рынков

Запускает пайплайн для всех реализованных рынков и сводит метрики в одну таблицу для сопоставления.

**Что делает:**
- Прогоняет `run_unified_pipeline` последовательно для USA (target: `CSCO`, 25 акций), Crypto (target: `BTC`, 25 монет) и Russia (target: `LKOH`, 25 бумаг MOEX)
- Собирает результаты в словарь `results` и передаёт в `compare_metrics_table`
- Выводит единую таблицу метрик бэктеста по всем рынкам рядом

**Когда использовать:** для финального сравнения качества стратегии на разных рынках после изменения параметров или данных. При добавлении нового рынка — добавить соответствующий блок в первую ячейку.

```python
from sarb.run import load_market_prices, load_run_config, run_unified_pipeline
from sarb.reporting import compare_metrics_table, display_metrics_table

results = {}

usa_config = load_run_config("usa")
usa_prices = load_market_prices("usa")
usa_target = "CSCO"
usa_universe = [symbol for symbol in usa_prices.columns if symbol != usa_target][:25]
results["usa"] = run_unified_pipeline(
    usa_prices,
    target=usa_target,
    universe=usa_universe,
    run_config=usa_config,
)

crypto_config = load_run_config("crypto")
crypto_prices = load_market_prices("crypto")
crypto_target = "BTC"
crypto_universe = [symbol for symbol in crypto_prices.columns if symbol != crypto_target][:25]
results["crypto"] = run_unified_pipeline(
    crypto_prices,
    target=crypto_target,
    universe=crypto_universe,
    run_config=crypto_config,
)

russia_config = load_run_config("russia")
russia_prices = load_market_prices("russia")
russia_target = "LKOH"
russia_universe = [symbol for symbol in russia_prices.columns if symbol != russia_target]
results["russia"] = run_unified_pipeline(
    russia_prices,
    target=russia_target,
    universe=russia_universe,
    run_config=russia_config,
)
```

```python
display_metrics_table(compare_metrics_table(results))
```
