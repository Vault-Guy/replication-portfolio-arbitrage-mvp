# Crypto unified pipeline

Hourly closes from `data/raw/crypto_data.zip` (2020-01-01 through 2026-01-01). Later listings leave incomplete histories before each coin's first bar.

**Market caveats:** crypto trades on a 24/7 calendar, liquidity and spreads vary by venue and regime, and the configured transaction costs may understate realized frictions.

```python
from sarb.reporting import display_metrics_table, plot_pipeline_dashboard
from sarb.run import (
    load_market_prices,
    load_run_config,
    metrics_dataframe,
    run_unified_pipeline,
)

run_config = load_run_config("crypto")
prices = load_market_prices("crypto")
target = "BTC"
universe = [symbol for symbol in prices.columns if symbol != target][:25]
result = run_unified_pipeline(
    prices,
    target=target,
    universe=universe,
    run_config=run_config,
)
```

```python
result.coverage

selection = {
    "target": result.target,
    "universe_size": len(result.universe),
    "universe": ", ".join(result.universe),
}
selection
```
