# 05 — Пайплайн: криптовалюты (Crypto)

Полный прогон стратегии репликации/арбитража на часовых данных криптовалютного рынка.

**Данные:** часовые цены закрытия из `data/raw/crypto_data.zip`, период 2020-01-01 — 2026-01-01. Монеты с более поздним листингом имеют неполную историю в начале периода.

**Настройки по умолчанию:**
- Target: `BTC`
- Universe: первые 25 символов из загруженных цен (исключая target)

**Что делает:**
- Загружает конфиг и цены через `load_run_config("crypto")` / `load_market_prices("crypto")`
- Запускает `run_unified_pipeline` — PCA-репликация, генерация сигналов, бэктест
- Выводит покрытие (`result.coverage`), состав юниверса и дашборд метрик

**Особенности рынка:** крипто торгуется 24/7, ликвидность и спреды существенно варьируются по биржам и режимам волатильности. Транзакционные издержки в конфиге могут занижать реальные затраты на исполнение.

```python
from sarb.reporting import display_metrics_table, plot_pipeline_dashboard
from sarb.run import (
    load_market_prices,
    load_run_config,
    metrics_dataframe,
    run_unified_pipeline,
)

run_config = load_run_config("crypto")
prices = load_market_prices("crypto")
target = "BTC"
universe = [symbol for symbol in prices.columns if symbol != target][:25]
result = run_unified_pipeline(
    prices,
    target=target,
    universe=universe,
    run_config=run_config,
)
```

```python
result.coverage

selection = {
    "target": result.target,
    "universe_size": len(result.universe),
    "universe": ", ".join(result.universe),
}
selection
```

```python
plot_pipeline_dashboard(result)
display_metrics_table(metrics_dataframe(result.backtest.metrics))
```
