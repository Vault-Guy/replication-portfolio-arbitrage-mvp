# 06 — Пайплайн: российские акции (Russia)

Полный прогон стратегии репликации/арбитража на дневных данных российского рынка (MOEX).

**Данные:** дневные цены закрытия из `data/russia_data.zip`, архив `LKOH_and_partners_data`, период 2013-02-19 — 2026-04-14. Всего 26 бумаг из индекса Мосбиржи.

**Настройки по умолчанию:**
- Target: `LKOH` (Лукойл — центральный актив архива)
- Universe: все остальные 25 бумаг

**Что делает:**
- Загружает конфиг и цены через `load_run_config("russia")` / `load_market_prices("russia")`
- Запускает `run_unified_pipeline` — PCA-репликация, генерация сигналов, бэктест
- Выводит покрытие (`result.coverage`), состав юниверса и дашборд метрик

**Особенности рынка:** данные охватывают периоды высокой волатильности (2014, 2020, 2022). Торговые дни MOEX отличаются от NYSE; транзакционные издержки выше из-за меньшей ликвидности ряда бумаг.

```python
from sarb.reporting import display_metrics_table, plot_pipeline_dashboard
from sarb.run import (
    load_market_prices,
    load_run_config,
    metrics_dataframe,
    run_unified_pipeline,
)

run_config = load_run_config("russia")
prices = load_market_prices("russia")
target = "LKOH"
universe = [symbol for symbol in prices.columns if symbol != target]
result = run_unified_pipeline(
    prices,
    target=target,
    universe=universe,
    run_config=run_config,
)
```

```python
result.coverage

selection = {
    "target": result.target,
    "universe_size": len(result.universe),
    "universe": ", ".join(result.universe),
}
selection
```

```python
plot_pipeline_dashboard(result)
display_metrics_table(metrics_dataframe(result.backtest.metrics))
```
