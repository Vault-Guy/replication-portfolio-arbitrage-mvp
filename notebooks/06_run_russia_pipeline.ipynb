**Что делает сейчас:**
- Пытается загрузить данные; при `NotImplementedError` выводит сообщение и устанавливает `prices = None`
- Все downstream-ячейки защищены проверкой `if prices is not None` и безопасно пропускаются

```python
from sarb.reporting import display_metrics_table, plot_pipeline_dashboard
from sarb.run import (
    load_market_prices,
    load_run_config,
    metrics_dataframe,
    run_unified_pipeline,
)

run_config = load_run_config("russia")
try:
    prices = load_market_prices("russia")
except NotImplementedError as exc:
    print(exc)
    prices = None
```

```python
if prices is None:
    print("TODO: set target and universe after the Russia loader is implemented.")
else:
    target = prices.columns[0]
    universe = [symbol for symbol in prices.columns if symbol != target][:25]
    result = run_unified_pipeline(
        prices,
        target=target,
        universe=universe,
        run_config=run_config,
    )
    result.coverage
    {
        "target": result.target,
        "universe_size": len(result.universe),
        "universe": ", ".join(result.universe),
    }
```

```python
if prices is not None:
    plot_pipeline_dashboard(result)
    display_metrics_table(metrics_dataframe(result.backtest.metrics))
```
