# US equities unified pipeline

Daily adjusted closes from `data/raw/usa_data.zip` (2016-01-01 through 2026-01-01).

**Future-selection bias:** the universe is the union of S&P 500 names in the top-100 by capitalization in 2016 and 2026. Using 2026 membership when studying earlier years can inflate backtest quality.

```python
from sarb.reporting import display_metrics_table, plot_pipeline_dashboard
from sarb.run import (
    load_market_prices,
    load_run_config,
    metrics_dataframe,
    run_unified_pipeline,
)

run_config = load_run_config("usa")
prices = load_market_prices("usa")
target = "CSCO"
universe = [symbol for symbol in prices.columns if symbol != target][:25]
result = run_unified_pipeline(
    prices,
    target=target,
    universe=universe,
    run_config=run_config,
)
```
