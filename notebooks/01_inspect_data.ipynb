# 01 — Инспекция сырых архивов

Проверяет содержимое сырых ZIP-архивов с рыночными данными перед загрузкой в пайплайн.

**Что делает:**
- Читает `data/usa_data.zip`, `data/crypto_data.zip` и `data/russia_data.zip` через `sarb.data.inspect_zip`
- Для каждого актива выводит: количество строк, пропуски, временной диапазон, основную частоту баров и количество нерегулярных гэпов
- Строит сводную таблицу по архивам: число активов, минимум/максимум строк, суммарные пропуски, крайние даты

**Когда использовать:** при добавлении нового архива или подозрении на повреждённые / неполные данные.

```python
from pathlib import Path

from sarb.config import PROJECT_ROOT
from sarb.data.inspect_zip import format_report, inspect_archives

archives = [
    PROJECT_ROOT / "data/usa_data.zip",
    PROJECT_ROOT / "data/crypto_data.zip",
    PROJECT_ROOT / "data/russia_data.zip",
]
reports = inspect_archives(archives)
for report in reports:
    print(format_report(report))
    print()
```

```python
import pandas as pd

summary_rows = []
for report in reports:
    for asset in report.assets:
        summary_rows.append(
            {
                "archive": Path(report.archive).name,
                "symbol": asset.symbol,
                "rows": asset.row_count,
                "missing": sum(asset.missing_by_column.values()),
                "first_timestamp": asset.first_timestamp,
                "last_timestamp": asset.last_timestamp,
                "primary_frequency": asset.primary_frequency,
                "frequency_mode_share": asset.frequency_mode_share,
                "irregular_gaps": asset.non_primary_gap_count,
            }
        )

summary = pd.DataFrame(summary_rows)
summary.groupby("archive").agg(
    assets=("symbol", "count"),
    min_rows=("rows", "min"),
    max_rows=("rows", "max"),
    total_missing=("missing", "sum"),
    earliest_start=("first_timestamp", "min"),
    latest_end=("last_timestamp", "max"),
)
```
